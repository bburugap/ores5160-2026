1. Count Vowels

Write a function `count_vowels(s)` that takes a string and returns the number of vowels in it (`a`, `e`, `i`, `o`, `u` — both upper and lower case).

```python
>>> count_vowels("hello")
2
>>> count_vowels("HEllO World")
3

````python
### BEGIN SOLUTION
def count_vowels(vow):
    """
    (str) -> int
    Returns the number of vowels in the string (case-insensitive).

    >>> count_vowels("hello")
    2
    >>> count_vowels("HEllO World")
    3
    >>> count_vowels("xyz")
    0
    """
    vowels = "aeiouAEIOU"
    count = 0

    for char in vow:
        if char in vowels:
            count += 1

    return count

### END SOLUTION
````

````python
assert count_vowels("hello") == 2
assert count_vowels("HELLO") == 2
assert count_vowels("xyz") == 0

````

2 BMI Calculator

Write a function bmi(weight_kg, height_m) that returns the Body Mass Index (BMI) using the formula:

$
BMI = \frac{\text{weight (kg)}}{(\text{height (m)})^2}
$

Include doctests for values like:
	•	weight_kg=70, height_m=1.75 → 22.86 (rounded to two decimals)

````python
### BEGIN SOLUTION
def bmi(weight_kg, height_m):
    """
    (float, float) -> float
    Returns the BMI rounded to 2 decimals using weight in kg and height in meters.

    >>> round(bmi(70, 1.75), 2)
    22.86
    >>> round(bmi(50, 1.6), 2)
    19.53
    >>> round(bmi(70, 0))
    0
    >>> round(bmi('120', 1.75))
    0
    """
    try:
        # Check for invalid values
        if height_m <= 0 or weight_kg <= 0:
            raise ValueError("Inputs must be greater than 0")

        result = weight_kg / (height_m ** 2)

    except ZeroDivisionError:
        result = 0

    except:
        result = 0

    return result
### END SOLUTION
````

````python
assert round(bmi(70, 1.75), 2) == 22.86
assert round(bmi(50, 1.6), 2) == 19.53
assert round(bmi(70,0),2) == 0
assert round(bmi('120',1.75),2) == 0
````

````python
import doctest
doctest.run_docstring_examples(bmi, globals(), verbose=True)
````

3 Filter Short Words

Create a function filter_short(words, max_len) that returns a list of only those words whose length is less than or equal to max_len.
```python
>>> filter_short(['hi', 'hello', 'no', 'yes'], 2)
['hi', 'no']

````python
def filter_short(words, max_len):
    """
    (list, int) -> list
    Returns a list of words with length <= max_len parameter.

    >>> filter_short(['hi', 'hello', 'no', 'yes'], 2)
    ['hi', 'no']
    >>> filter_short(['apple', 'banana', 'pear'], 4)
    ['pear']
    """
    result = []

    for word in words:
        if len(word) <= max_len:
            result.append(word)

    return result
````

````python
assert filter_short(['hi', 'hello', 'no', 'yes'], 2) == ['hi', 'no']
assert filter_short(['apple', 'banana', 'pear'], 4) == ['pear']

````
